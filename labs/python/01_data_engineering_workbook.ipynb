## 1. SQLite — default local database

SQLite needs no server or account. Use it for foundations, joins, aggregations, transactions, and Python database access.

```python
sqlite_conn = get_connection('sqlite')
query_rows(sqlite_conn, '''
SELECT c.country, COUNT(*) AS completed_orders,
       ROUND(SUM(ot.order_total), 2) AS revenue
FROM order_totals AS ot
JOIN customers AS c ON c.customer_id = ot.customer_id
WHERE ot.status = 'completed'
GROUP BY c.country
ORDER BY revenue DESC
''')
```
